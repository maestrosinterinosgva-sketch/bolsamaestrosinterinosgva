### 1️⃣ Paso 1: Conectar con GitHub y preparar el entorno
Introduce tu **Token de GitHub** para que el bot pueda publicar directamente los cambios en tus páginas webs:

```python
#@title 🔑 Conexión con GitHub y Preparación
import os

GITHUB_TOKEN = "" #@param {type:"string"}

print("[*] [1/3] Instalando dependencias necesarias...")
!pip install -q pymupdf

print("[*] [2/3] Conectando con los repositorios de GitHub...")
REPO_BOLSA = "https://github.com/maestrosinterinosgva-sketch/bolsamaestrosinterinosgva.git"
REPO_DESTINOS = "https://github.com/maestrosinterinosgva-sketch/destinosinterinosgva.git"

auth_bolsa = REPO_BOLSA.replace("https://", f"https://{GITHUB_TOKEN}@") if GITHUB_TOKEN else REPO_BOLSA
auth_destinos = REPO_DESTINOS.replace("https://", f"https://{GITHUB_TOKEN}@") if GITHUB_TOKEN else REPO_DESTINOS

# 1. Clonar o actualizar repositorio principal (Maestros y Secundaria)
if not os.path.exists("bolsa"):
    !git clone {auth_bolsa} bolsa
else:
    %cd bolsa
    !git pull
    %cd ..

# 2. Clonar o actualizar repositorio de Destinos Primaria
if not os.path.exists("destinos"):
    !git clone {auth_destinos} destinos
else:
    %cd destinos
    !git pull
    %cd ..

print("[OK] [3/3] Entorno y repositorios listos para actualizar.")
```
